# Conhecimento

### O que é o duelo de modelos em machine learning?
O duelo de modelos em machine learning refere-se à comparação prática de diferentes algoritmos de aprendizado de máquina para determinar qual deles oferece o melhor desempenho em um determinado conjunto de dados. No contexto do Módulo 41, isso envolve a comparação entre SVM (Support Vector Machine) e XGBoost usando a base de dados do Titanic.

### Quais são as principais características do SVM e do XGBoost?
O SVM é conhecido por encontrar a melhor fronteira de decisão entre classes, mesmo em dados não linearmente separáveis, enquanto o XGBoost é reconhecido por sua precisão e capacidade de lidar com grandes volumes de dados através de um processo de ensemble. Ambos os modelos têm suas vantagens e desvantagens, especialmente em relação ao ajuste de hiperparâmetros e recursos computacionais.

### Por que o XGBoost inicialmente supera o SVM em precisão?
O XGBoost inicialmente supera o SVM em precisão devido à sua capacidade de ensemble, que combina múltiplas árvores de decisão para melhorar a precisão do modelo. No entanto, é importante considerar o potencial de overfitting dos modelos baseados em árvores, que pode afetar o desempenho em dados não vistos.

### Como o pré-processamento de dados afeta o desempenho dos modelos?
O pré-processamento de dados, incluindo padronização e balanceamento, pode melhorar significativamente o desempenho dos modelos. No caso do SVM, essas técnicas são particularmente benéficas, especialmente em problemas onde a previsão correta da classe minoritária é crítica.

### O que é Grid Search e como ele é usado na otimização de hiperparâmetros?
Grid Search é uma técnica de otimização de hiperparâmetros que testa todas as combinações possíveis de parâmetros para encontrar a melhor configuração para um modelo. No XGBoost, isso pode envolver ajustes na profundidade máxima da árvore e na taxa de aprendizado, enquanto no SVM, diferentes valores de C e kernels são testados.

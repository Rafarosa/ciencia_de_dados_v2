# Glossário

### Balanceamento de classes
Técnica utilizada para ajustar a distribuição das classes em um conjunto de dados, especialmente em problemas de classificação onde uma classe pode ser significativamente menor que as outras, afetando o desempenho do modelo.

### Ensemble
Método que combina múltiplos modelos de aprendizado para melhorar a precisão e a robustez das previsões, como é o caso do XGBoost, que utiliza árvores de decisão em conjunto.

### Fronteira de decisão
Limite que separa diferentes classes em um espaço de características, utilizado por algoritmos como o SVM para classificar dados de forma eficaz.

### Grid Search
Técnica de otimização de hiperparâmetros que testa exaustivamente todas as combinações possíveis de parâmetros para encontrar a configuração que proporciona o melhor desempenho do modelo.

### Kernel RBF
Função de kernel utilizada no SVM que transforma dados em um espaço de maior dimensão, permitindo a separação de classes que não são linearmente separáveis no espaço original.

### Overfitting
Fenômeno onde um modelo de machine learning se ajusta excessivamente aos dados de treinamento, capturando ruídos e padrões irrelevantes, o que prejudica seu desempenho em dados novos.

### Padronização
Processo de escalonar características de dados para que tenham média zero e desvio padrão um, o que pode melhorar o desempenho de modelos de machine learning, especialmente o SVM.

### Taxa de aprendizado
Parâmetro que controla o tamanho dos passos dados na direção do mínimo durante o treinamento de um modelo, influenciando a velocidade e a estabilidade da convergência.

### Validação cruzada
Técnica de avaliação de modelos que divide o conjunto de dados em múltiplas partes, treinando e testando o modelo em diferentes subconjuntos para garantir que o desempenho seja consistente.

### XGBoost
Algoritmo de machine learning baseado em árvores de decisão que utiliza técnicas de ensemble para melhorar a precisão e a eficiência, sendo altamente eficaz em competições de dados.

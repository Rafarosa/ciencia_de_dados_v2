# Glossário

### AdaBoost
Um método de Boosting que ajusta o peso dos modelos de forma iterativa, dando mais importância aos exemplos que foram classificados incorretamente nas iterações anteriores, melhorando assim a precisão do modelo.

### Bagging
Também conhecido como Bootstrap Aggregating, é uma técnica de ensemble que treina múltiplos modelos em diferentes subconjuntos dos dados e combina suas previsões para reduzir a variância e melhorar a precisão.

### Boosting
Uma técnica de ensemble que cria uma sequência de modelos, onde cada novo modelo tenta corrigir os erros dos anteriores, resultando em alta precisão e redução de viés.

### Gradient Boosting
Um método de Boosting que otimiza a função de perda através da adição de modelos sequenciais, cada um corrigindo os erros residuais dos modelos anteriores.

### Grid Search
Uma técnica de otimização de hiperparâmetros que realiza uma busca exaustiva através de um espaço especificado de parâmetros, avaliando todas as combinações possíveis para encontrar a melhor configuração.

### Hiperparâmetros
Parâmetros externos ao modelo que precisam ser definidos antes do treinamento, influenciando diretamente o desempenho e a eficácia dos modelos de ensemble.

### Random Forest
Um método de Bagging que utiliza múltiplas árvores de decisão treinadas em diferentes subconjuntos dos dados para melhorar a precisão e reduzir a variância das previsões.

### Random Search
Uma técnica de otimização de hiperparâmetros que seleciona aleatoriamente combinações de parâmetros a partir de um espaço definido, oferecendo uma alternativa mais eficiente ao Grid Search.

### XGBoost
Uma implementação avançada de Gradient Boosting que é altamente eficiente e escalável, conhecida por sua capacidade de lidar com grandes volumes de dados e oferecer alta precisão nas previsões.

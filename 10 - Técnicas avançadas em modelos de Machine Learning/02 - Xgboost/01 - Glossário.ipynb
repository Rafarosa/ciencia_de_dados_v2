# Glossário

### Boosting
Uma técnica de ensemble que combina múltiplos modelos fracos para criar um modelo mais robusto e preciso, melhorando a performance preditiva.

### Gradient Boosting
Um método de boosting que utiliza árvores de decisão para otimizar previsões, eficaz tanto para tarefas de classificação quanto de regressão.

### L1 Regularization (Lasso)
Uma técnica de regularização que adiciona uma penalidade equivalente ao valor absoluto da magnitude dos coeficientes, ajudando a evitar o overfitting.

### L2 Regularization
Uma técnica de regularização que adiciona uma penalidade equivalente ao quadrado da magnitude dos coeficientes, também conhecida como Ridge Regression.

### Overfitting
Um problema em que o modelo se ajusta excessivamente aos dados de treinamento, perdendo a capacidade de generalizar para novos dados.

### Parada Antecipada
Uma técnica que interrompe o treinamento do modelo quando o desempenho em um conjunto de validação começa a piorar, evitando overfitting e reduzindo o tempo de treinamento.

### Regularização
Um conjunto de técnicas utilizadas para melhorar a capacidade de generalização do modelo, evitando que ele se ajuste excessivamente aos dados de treinamento.

### Seleção de Variáveis
O processo de escolher quais variáveis incluir no modelo, avaliando o impacto dessa escolha nos resultados do modelo.

### XGBoost
Uma implementação avançada do algoritmo de boosting, conhecida por sua eficiência, flexibilidade e performance, com técnicas de regularização para evitar overfitting.

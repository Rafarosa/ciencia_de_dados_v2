### D) Verifique se os dados das variáveis Booleanas são balanceados ou não.

As variáveis booleanas nesta amostra são:
- Elderly
- Married
- Dependents
- PhoneService
- Security_Service
- Technical_Support
- StreamingTV
- Churn

Já verificamos os dados de 'Dependents' e 'Churn' e podemos dizer que essas variáveis não são balanceadas

Para facilitar o processo, vamos criar uma função para gerar os gráficos.

Observando os gráficos gerados, apenas a variável 'Married' apresenta um grau de balanceamento elevado. Nenhuma das outra variáveis estão balanceadas
Majoritariamente temos:
- Pessoas mais com idade inferior a 60 anos
- Possuem serviço telefônico
- Não possuem serviço de segurança contratado
- Já utilizaram, em algum momento, o suporte técnico da empresa
- Não possuem o serviço de Streaming do provedor 


```python
def plot_distribuicao_boll(df, column_name):
    """
    Gera um gráfico de barras mostrando a distribuição de uma variável booleana no dataframe.

    Parâmetros:
    df (pd.DataFrame): DataFrame contendo a variável booleana.
    column_name (str): Nome da variável booleana a ser analisada.

    Retorno:
    Exibe um gráfico de barras com a distribuição da variável selecionada.
    """
    # Contagem de valores
    count_values = df[column_name].value_counts()
    print(f'Contagem de {column_name}:\n{count_values}')
    
    # Calcular a porcentagem
    percent_values = (count_values / count_values.sum()) * 100
    print(f'Percentual de {column_name}:\n{percent_values}')
    
    # Criando gráfico de barras
    fig = px.bar(
        count_values, 
        x=count_values.index, 
        y=count_values.values
    )
    
    fig.update_layout(
        title=f'Distribuição de {column_name}',
        xaxis_title=column_name,
        yaxis_title='Contagem',
        xaxis=dict(tickvals=[0, 1], ticktext=['Não', 'Sim']),
        yaxis=dict(title='Contagem'),
        showlegend=False
    )
    
    fig.show()
```

```python
plot_distribuicao_boll(df, "Elderly")
plot_distribuicao_boll(df, "Married")
plot_distribuicao_boll(df, "PhoneService")
plot_distribuicao_boll(df, "Security_Service")
plot_distribuicao_boll(df, "Technical_Support")
plot_distribuicao_boll(df, "StreamingTV")
```

Avaliação

Apesar de apresentar um desvio padrão elevado, não forma considerados outliers na avaliação dos graficos de box plot.
Isso de deve pela natureza dos dados que podem variar, como o tempo de permanencia do cliente com algum vinculo e o pagamento mensal.
O Total pago também leva em consideração o valor pago pelo tempo de contrato.
Neste caso vamos manter as os dados como estão.


```python
#Seu código aqui
def plot_box_outliers(df, columns_name):
    """
    Plota boxplots para as colunas especificadas em um DataFrame.

    Args:
        df (pd.DataFrame): DataFrame contendo os dados.
        columns_name (list): Lista de nomes das colunas a serem plotadas.

    Returns:
        None
    """
    # montando o gráfico com plotly
    fig = px.box(df, y=columns_name)
    fig.update_layout(
        title=f'Boxplot de {columns_name}',
        yaxis_title=columns_name,
        xaxis_title='Valores',
        showlegend=False
    )
    # Exibindo o gráfico
    fig.show()

plot_box_outliers(df,"Time_as_a_customer")
plot_box_outliers(df,"Monthly_Payment")
plot_box_outliers(df,"Total_Paid")
```

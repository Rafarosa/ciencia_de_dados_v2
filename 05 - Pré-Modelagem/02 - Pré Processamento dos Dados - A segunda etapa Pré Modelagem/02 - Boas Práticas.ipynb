### Conheça profundamente a base de dados
Antes de iniciar qualquer análise, é crucial entender a base de dados em profundidade. Isso inclui conhecer o tipo de variável, colunas, registros e tipos de dados. Tornar-se um especialista no assunto abordado pela base de dados permite fazer perguntas relevantes e obter insights valiosos.

### Realize análise univariada
A análise univariada é fundamental para entender a distribuição de cada variável individualmente. Utilize estatísticas descritivas para variáveis quantitativas e categóricas para identificar padrões e anomalias. Isso ajuda a detectar outliers e a compreender a distribuição dos dados.

### Identifique e trate outliers
Outliers podem distorcer análises e modelos preditivos. Identifique-os utilizando métodos estatísticos e gráficos, e trate-os de maneira adequada, seja removendo, substituindo ou transformando os dados. A decisão deve ser baseada em uma análise detalhada e conhecimento profundo dos dados.

### Realize análise bivariada
A análise bivariada é essencial para entender a relação entre duas variáveis. Isso pode revelar padrões e correlações importantes que são úteis para modelagem preditiva e tomada de decisões estratégicas. Utilize gráficos e visuais para interpretar e comunicar esses insights.

### Exemplo de utilização no mercado de trabalho:
 - Segmentação de Clientes: Empresas utilizam análise univariada e bivariada para segmentar clientes com base em características demográficas e comportamentais, permitindo campanhas de marketing mais direcionadas e eficazes.

 - Detecção de Fraudes: A identificação de outliers em transações financeiras pode ajudar a detectar atividades fraudulentas. Análises univariadas e bivariadas são usadas para identificar padrões incomuns que podem indicar fraude.

## Exemplo de código executável (em Python) que ilustra a análise univariada e detecção de outliers:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar dados
df = pd.read_csv('dados.csv')

# Análise univariada
print(df['balanco'].describe())

# Detecção de outliers usando boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['balanco'])
plt.title('Boxplot do Balanço')
plt.show()

# Tratamento de outliers - Remoção
Q1 = df['balanco'].quantile(0.25)
Q3 = df['balanco'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

df_sem_outliers = df[(df['balanco'] >= limite_inferior) & (df['balanco'] <= limite_superior)]
print(df_sem_outliers['balanco'].describe())
```

Vamos discutir o código passo-a-passo:

import pandas as pd: Importa a biblioteca pandas para manipulação de dados.

import matplotlib.pyplot as plt: Importa a biblioteca matplotlib para visualização de dados.

import seaborn as sns: Importa a biblioteca seaborn para visualização de dados estatísticos.

df = pd.read_csv('dados.csv'): Carrega os dados de um arquivo CSV para um DataFrame.

print(df['balanco'].describe()): Exibe estatísticas descritivas da variável 'balanco'.

sns.boxplot(x=df['balanco']): Cria um boxplot para visualizar a distribuição da variável 'balanco' e identificar outliers.

Q1 = df['balanco'].quantile(0.25): Calcula o primeiro quartil (Q1) da variável 'balanco'.

Q3 = df['balanco'].quantile(0.75): Calcula o terceiro quartil (Q3) da variável 'balanco'.

IQR = Q3 - Q1: Calcula o intervalo interquartil (IQR) da variável 'balanco'.

limite_inferior = Q1 - 1.5 * IQR: Define o limite inferior para detecção de outliers.

limite_superior = Q3 + 1.5 * IQR: Define o limite superior para detecção de outliers.

df_sem_outliers = df[(df['balanco'] >= limite_inferior) & (df['balanco'] <= limite_superior)]: Remove os outliers da variável 'balanco' com base nos limites definidos.

print(df_sem_outliers['balanco'].describe()): Exibe estatísticas descritivas da variável 'balanco' após a remoção dos outliers.

# Conhecimento

### O que é o comando DISTINCT em SQL e quando devo usá-lo?
O comando DISTINCT em SQL é utilizado para selecionar valores únicos em uma coluna, eliminando duplicatas. Ele é especialmente útil em bases de dados volumosas quando você precisa identificar ou contar valores distintos em um conjunto de dados.

### Como o comando WHERE é utilizado em consultas SQL?
O comando WHERE é utilizado em consultas SQL para filtrar dados com base em condições específicas. Ele permite o uso de operadores como LIKE, =, IN, NOT IN, e NOT LIKE para definir critérios de seleção, tornando as consultas mais precisas e eficientes.

### Qual a importância da sensibilidade a maiúsculas e minúsculas em SQL?
A sensibilidade a maiúsculas e minúsculas em SQL é importante porque pode afetar o resultado das consultas, especialmente ao comparar strings. Funções como LOWER e UPPER são usadas para realizar comparações de forma insensível a maiúsculas, garantindo que as consultas retornem resultados consistentes.

### Como posso formatar consultas SQL para facilitar a leitura?
Para facilitar a leitura de consultas SQL, é recomendado usar boas práticas de escrita e formatação, como a utilização de aspas triplas para permitir a quebra de linhas. Isso ajuda a manter o código organizado e mais fácil de entender e manter.

### Quais são os operadores lógicos mais comuns em SQL e como são usados?
Os operadores lógicos mais comuns em SQL são AND, OR, NOT, IS NULL e IS NOT NULL. Eles são usados para realizar filtragens complexas e precisas em bases de dados, permitindo a combinação de múltiplas condições em uma única consulta.

### Para que servem as cláusulas ORDER BY e GROUP BY em SQL?
A cláusula ORDER BY é utilizada para ordenar os resultados de uma consulta, enquanto a cláusula GROUP BY permite agrupar linhas que compartilham valores em comum. GROUP BY é frequentemente usado com funções agregadas como SUM, AVG, MAX e MIN para realizar cálculos em grupos de dados.

### Como posso estruturar meu pensamento antes de escrever uma query SQL?
Antes de escrever uma query SQL, é útil listar as colunas necessárias, filtros, agrupamentos e ordenações. Essa abordagem sistemática ajuda a clarificar o que precisa ser feito e onde cada elemento deve ser colocado na query, facilitando a construção de consultas eficientes e precisas.

### Como as funções LOWER e UPPER são usadas em SQL?
As funções LOWER e UPPER são usadas em SQL para converter strings para minúsculas ou maiúsculas, respectivamente. Elas são essenciais para realizar comparações de strings de forma insensível a maiúsculas, simplificando o processo de filtragem de dados.

### Quais são algumas boas práticas ao escrever queries SQL?
Algumas boas práticas ao escrever queries SQL incluem: usar indentação e quebras de linha para melhorar a legibilidade, comentar o código para explicar partes complexas, evitar o uso de SELECT *, e sempre testar as queries em um ambiente seguro antes de aplicá-las em produção.

# Glossário

### Cláusula GROUP BY
Utilizada para agrupar linhas que compartilham valores em comum em uma consulta SQL, frequentemente em conjunto com funções agregadas como SUM, AVG, MAX e MIN.

### Cláusula ORDER BY
Usada para ordenar os resultados de uma consulta SQL com base em uma ou mais colunas, podendo ser em ordem ascendente ou descendente.

### Comando DISTINCT
Um comando SQL que seleciona valores únicos em uma coluna, eliminando duplicatas, útil em bases de dados volumosas.

### Função LOWER
Uma função SQL que converte todos os caracteres de uma string para minúsculas, facilitando comparações insensíveis a maiúsculas.

### Função UPPER
Uma função SQL que converte todos os caracteres de uma string para maiúsculas, utilizada para comparações insensíveis a maiúsculas.

### Operador LIKE
Um operador SQL utilizado para buscar um padrão específico dentro de uma coluna de texto, permitindo o uso de curingas como % e _.

### Operador NOT LIKE
Um operador SQL que busca registros que não correspondem a um padrão específico dentro de uma coluna de texto.

### Operador IS NULL
Um operador SQL usado para verificar se um valor em uma coluna é nulo, indicando a ausência de dados.

### Operador IS NOT NULL
Um operador SQL que verifica se uma coluna contém um valor não nulo, garantindo que a coluna tenha dados.

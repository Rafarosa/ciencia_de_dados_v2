# Conhecimento

### O que é um banco de dados relacional?
Um banco de dados relacional é um tipo de banco de dados que organiza dados em tabelas, que podem ser relacionadas entre si. Ele utiliza a linguagem SQL para gerenciar e consultar dados, garantindo integridade e suporte a transações complexas.

### Qual é a diferença entre bancos de dados relacionais e não relacionais?
Bancos de dados relacionais têm uma estrutura rígida e usam SQL para manipulação de dados, sendo ideais para aplicações que requerem integridade de dados. Bancos de dados não relacionais (NoSQL) são mais flexíveis e adequados para grandes volumes de dados não estruturados, como redes sociais e Big Data.

### O que são SGBDs e por que são importantes?
Sistemas de Gerenciamento de Bancos de Dados (SGBDs) são softwares que permitem a criação, manipulação e administração de bancos de dados. Eles são importantes porque facilitam o gerenciamento eficiente de grandes volumes de dados, garantindo segurança, integridade e acesso controlado.

### Como o SQL é integrado ao Python no Google Colab?
No Google Colab, o SQL pode ser integrado ao Python usando bibliotecas específicas que permitem a execução de consultas SQL diretamente de scripts Python. Isso facilita a análise de dados, combinando a capacidade de consulta do SQL com o poder de processamento do Python.

### Quais são os comandos básicos do SQL e suas funções?
Os comandos básicos do SQL incluem SELECT, que é usado para buscar dados de uma tabela, e FROM, que especifica a tabela de onde os dados devem ser extraídos. Esses comandos são essenciais para visualizar e manipular dados em um banco de dados.

### Por que é importante renomear colunas em consultas SQL?
Renomear colunas em consultas SQL é importante para tornar os resultados mais legíveis e compreensíveis. Isso é especialmente útil quando se realizam operações matemáticas ou agregações, onde os nomes originais das colunas podem não refletir adequadamente o conteúdo dos dados resultantes.

### Como o SQL pode ser usado para realizar operações matemáticas?
O SQL pode realizar operações matemáticas diretamente no banco de dados usando funções como AVG para calcular médias e SUM para somar valores de colunas. Isso permite que cálculos sejam feitos de forma eficiente sem precisar exportar os dados para outro ambiente.

### Qual é a importância do pré-processamento de dados?
O pré-processamento de dados é crucial para garantir que os dados estejam limpos, consistentes e prontos para análise e modelagem. Ele envolve etapas como limpeza, transformação e normalização dos dados, o que ajuda a melhorar a qualidade dos resultados analíticos.

### O que são data warehouses na nuvem e quais são suas vantagens?
Data warehouses na nuvem, como Google BigQuery e Amazon Redshift, são plataformas que permitem armazenar e processar grandes volumes de dados de forma escalável e eficiente. Suas vantagens incluem flexibilidade, escalabilidade, e a capacidade de lidar com grandes conjuntos de dados sem a necessidade de infraestrutura física.

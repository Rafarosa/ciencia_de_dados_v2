Para nossa atividade inicial de SQL trabalharemos com uma base de vendas.
Abaixo temos o ambiente de preparo para trabalharmos com o SQL lite.

```python
import sqlite3 #Importando a biblioteca necessária
import pandas as pd
```

```python
df_vendas = pd.read_csv("TB_VENDAS_TAREFA.csv", delimiter=';')
```

```python
conn = sqlite3.connect(':memory:')
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace')
```

```python
def run_query(query):
    return pd.read_sql_query(query, conn)
```

# 1) Faça uma consulta que retorne todos os dados da nossa tabela de vendas, com todas linhas e todas colunas.


```python
query = "SELECT * FROM tb_vendas;"
result_df = run_query(query)
print(result_df)
```

# 2) Faça uma consulta que retorne apenas 10 linhas da coluna produto.


```python
query = "SELECT * FROM tb_vendas LIMIT 10"
result_df = run_query(query)
print(result_df)
```

# 3) Faça uma consulta que retorne a média do valor por unidade e também das unidades vendidas. Não se esqueça de renomear o nome das colunas ao cria-las.

```python
query = "SELECT AVG(CAST(REPLACE(VALOR_UNID, ',', '.') AS FLOAT)) AS MEDIA_VALOR_UNID, AVG(UNIDADES) AS MEDIA_UNIDADE FROM tb_vendas"
result_df = run_query(query)
print(result_df)
```

# 4) Note que na nossa tabela não temos o total de valor gasto por compra, temos o valor do produto e a quantidade vendida, para termos o valor total precisamos multiplicar essas duas colunas e para multiplicar colunas usamos o *.
# Faça uma consulta que retorne a coluna ID_COMPRA, ID_CLIENTE e o valor total gasto.

Ex: SELECT TICKET_MEDIO*UNIDADE FROM TB_EXEMPLO

```python
query = "SELECT ID_COMPRA, ID_CLIENTE, (CAST(REPLACE(VALOR_UNID, ',', '.') AS FLOAT) * UNIDADES) AS VALOR_TOTAL_GASTO FROM tb_vendas;"
result_df = run_query(query)
print(result_df)
```

# 5) [DESAFIO] Faça uma consulta que traz a média do valor total gasto que foi calculado no exercício anterior. Note que não queremos a média por pedido e sim a média total, apenas essa informação deve vir na consulta.

```python
query = "SELECT AVG(CAST(REPLACE(VALOR_UNID, ',', '.') AS FLOAT) * UNIDADES) AS VALOR_TOTAL_GASTO FROM tb_vendas;"
result_df = run_query(query)
# Exibir o resultado
print(result_df)
```

# 7) Diferencie com suas palavras banco de dados relacionais de não relacionais.

**Resposta:**

Podemos diferenciar os banco de dados relacionais e não relacionais pelas suas características.
Bancos de dados relacionais são organizados com dados em tabelas com linhas e colunas e tem uma relação muito forte com integridade dos dados, trazendo com sigo uma rigidez estrutural. Esse conceito de estrutura de dados é ideal para dados que necessitam de consistência transacional.
Já o Banco de dados não relacional (Nosql), Não há tabelas fixas. Os dados são armazenados em formatos mais flexíveis, como pares de chave-valor, documentos (JSON), grafos ou colunas largas. Isso permite que você armazene dados de diferentes tipos no mesmo "documento". Alta escalabilidade, flexibilidade para lidar com mudanças no esquema de dados e desempenho rápido para grandes volumes de dados são as suas principais vantagens.

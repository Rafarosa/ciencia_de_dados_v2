# Glossário

### Data Warehouse
Um sistema de armazenamento de dados projetado para consultas e análises rápidas, que agrega dados de diferentes fontes para facilitar a geração de relatórios e insights.

### Google BigQuery
Um serviço de data warehouse na nuvem oferecido pelo Google, que permite executar consultas SQL em grandes volumes de dados de forma rápida e escalável.

### Integridade de Dados
O princípio de garantir que os dados em um banco de dados sejam precisos, consistentes e confiáveis ao longo de seu ciclo de vida.

### NoSQL
Um tipo de banco de dados que não utiliza a estrutura tabular dos bancos de dados relacionais, sendo mais flexível e adequado para dados não estruturados e grandes volumes de dados.

### Pré-processamento de Dados
A etapa de preparação dos dados antes da análise, que inclui limpeza, transformação e organização para garantir que estejam prontos para modelagem e análise.

### Renomear Colunas
A prática de alterar os nomes das colunas em um conjunto de dados para torná-los mais descritivos e amigáveis, facilitando a interpretação dos resultados.

### SGBD (Sistema de Gerenciamento de Banco de Dados)
Software que permite a criação, manipulação e administração de bancos de dados, garantindo a integridade e segurança dos dados armazenados.

### SQL (Structured Query Language)
Uma linguagem de programação utilizada para gerenciar e manipular bancos de dados relacionais, permitindo a execução de consultas, inserções, atualizações e exclusões de dados.

### Transações Complexas
Operações que envolvem múltiplas etapas em um banco de dados, garantindo que todas as etapas sejam concluídas com sucesso ou que nenhuma seja aplicada, mantendo a integridade dos dados.

### Visualização de Dados
A representação gráfica de dados para facilitar a compreensão e análise, permitindo que padrões e insights sejam identificados de forma mais intuitiva.

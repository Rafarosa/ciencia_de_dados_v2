# Glossário

### Chave Primária
Um campo ou conjunto de campos em uma tabela que identifica exclusivamente cada registro nessa tabela. É essencial para a integridade dos dados e para realizar junções eficazes.

### Chave Estrangeira
Um campo em uma tabela que é a chave primária em outra tabela, usada para criar uma relação entre as duas tabelas e permitir junções de dados.

### Comando CREATE TABLE
Um comando SQL usado para criar uma nova tabela no banco de dados, especificando a estrutura e os tipos de dados para cada coluna.

### Comando INNER JOIN
Um tipo de junção SQL que retorna registros que têm valores correspondentes em ambas as tabelas envolvidas na junção.

### Comando LEFT JOIN
Um tipo de junção SQL que retorna todos os registros da tabela à esquerda e os registros correspondentes da tabela à direita. Se não houver correspondência, os resultados da tabela à direita serão nulos.

### Comando RIGHT JOIN
Um tipo de junção SQL que retorna todos os registros da tabela à direita e os registros correspondentes da tabela à esquerda. Se não houver correspondência, os resultados da tabela à esquerda serão nulos.

### Comando FULL JOIN
Um tipo de junção SQL que retorna todos os registros quando há uma correspondência em uma das tabelas. Se não houver correspondência, os resultados serão nulos para a tabela que não tem correspondência.

### Comando DISTINCT
Um comando SQL usado para remover duplicatas de um conjunto de resultados, garantindo que apenas valores únicos sejam retornados.

### Exportação para CSV
O processo de exportar dados de um banco de dados SQL para um arquivo CSV, que pode ser usado para análise em outras ferramentas ou para compartilhamento de dados.

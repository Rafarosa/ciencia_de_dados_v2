# Conhecimento

### Por que é importante aprender a criar tabelas no SQL?
Aprender a criar tabelas no SQL é importante porque permite que você defina a estrutura de armazenamento de dados de forma organizada e eficiente. Embora cientistas de dados geralmente trabalhem com tabelas existentes, entender como criar tabelas ajuda a compreender melhor a organização dos dados e a preparar o terreno para manipulações e análises mais complexas.

### O que é um "Join" e por que é essencial no SQL?
Um "Join" é uma operação no SQL que permite combinar dados de duas ou mais tabelas com base em uma coluna comum, como um ID ou CPF. É essencial porque, em ambientes corporativos, os dados são frequentemente distribuídos em várias tabelas, e os "Joins" permitem que você reúna essas informações para análises mais completas e significativas.

### Quais são os tipos de "Join" mais comuns e suas diferenças?
Os tipos de "Join" mais comuns são `INNER JOIN`, `LEFT JOIN`, `RIGHT JOIN` e `FULL JOIN`. `INNER JOIN` retorna apenas as linhas que têm correspondência em ambas as tabelas. `LEFT JOIN` retorna todas as linhas da tabela à esquerda e as correspondências da tabela à direita. `RIGHT JOIN` faz o oposto, retornando todas as linhas da tabela à direita e as correspondências da tabela à esquerda. `FULL JOIN` retorna todas as linhas quando há uma correspondência em uma das tabelas.

### Como a prática de "Join" pode ser aprimorada?
A prática de "Join" pode ser aprimorada através de exercícios contínuos e variados que desafiem a criar consultas complexas. Trabalhar com diferentes conjuntos de dados e cenários ajuda a tornar o uso de "Join" mais intuitivo e a desenvolver uma compreensão mais profunda de como combinar dados de maneira eficaz.

### Qual é a importância de calcular médias e aplicar filtros em consultas SQL?
Calcular médias e aplicar filtros em consultas SQL são operações fundamentais para a análise de dados. Elas permitem que você extraia insights significativos dos dados, como identificar tendências, comparar desempenhos e tomar decisões informadas com base em métricas específicas.

### Por que é importante evitar duplicações usando o comando `DISTINCT`?
Evitar duplicações usando o comando `DISTINCT` é importante para garantir que os resultados das consultas sejam precisos e não inflacionados por registros repetidos. Isso é crucial para análises corretas e para a apresentação de dados em relatórios e dashboards, onde a precisão é fundamental.

### Como o uso de ferramentas de visualização complementa o SQL?
O uso de ferramentas de visualização complementa o SQL ao permitir que os dados sejam apresentados de forma visualmente atraente e fácil de interpretar. Ferramentas como Looker Studio ou Power BI ajudam a transformar consultas SQL em gráficos e dashboards interativos, facilitando a comunicação de insights e a tomada de decisões baseadas em dados.

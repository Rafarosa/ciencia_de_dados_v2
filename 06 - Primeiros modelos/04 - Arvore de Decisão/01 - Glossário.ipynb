# Glossário

#### Acurácia
Uma métrica de avaliação de modelos de classificação que representa a proporção de previsões corretas em relação ao total de previsões feitas.

#### Entropia
Uma medida de impureza ou incerteza em um conjunto de dados, usada para determinar a melhor divisão em uma árvore de decisão.

#### Feature Importance
Uma técnica que quantifica a importância de cada variável de entrada em um modelo de árvore de decisão, ajudando a identificar quais variáveis têm maior impacto nas previsões.

#### Índice de Gini
Uma medida de impureza utilizada em árvores de decisão para avaliar a qualidade de uma divisão, sendo mais eficiente computacionalmente que a entropia.

#### Overfitting
Um problema em modelos de aprendizado de máquina onde o modelo se ajusta excessivamente aos dados de treinamento, perdendo a capacidade de generalizar para novos dados.

#### Poda
Uma técnica utilizada para reduzir o tamanho de uma árvore de decisão, removendo ramos que têm pouca importância, a fim de melhorar a generalização do modelo.

#### Precisão
Uma métrica de avaliação que indica a proporção de verdadeiros positivos em relação ao total de previsões positivas feitas pelo modelo.

#### Recall
Uma métrica de avaliação que indica a proporção de verdadeiros positivos em relação ao total de positivos reais no conjunto de dados.

#### Scikit-learn
Uma biblioteca de aprendizado de máquina em Python que fornece ferramentas simples e eficientes para análise de dados e modelagem preditiva.

#### Viés
A diferença entre a estimativa de um parâmetro feita pelo modelo e o valor real desse parâmetro, podendo levar a previsões sistematicamente incorretas.

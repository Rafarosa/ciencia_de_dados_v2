# Conhecimento

### O que é regressão linear e por que é importante?
A regressão linear é uma técnica de aprendizado supervisionado usada para prever valores e entender relações entre variáveis. Ela é importante porque permite modelar a relação entre uma variável dependente e uma ou mais variáveis independentes, facilitando previsões e análises de tendências.

### Qual é a diferença entre correlação e regressão?
Correlação mede a força e a direção de uma relação linear entre duas variáveis, enquanto a regressão quantifica e modela essa relação para fazer previsões. A regressão linear utiliza a equação da reta (Y = mX + b) para modelar a relação entre variáveis.

### Como a regressão linear simples é aplicada na prática?
A regressão linear simples é aplicada modelando a relação entre uma variável dependente e uma independente. Na prática, isso pode ser feito manualmente ou usando ferramentas como Python para automatizar cálculos. Um exemplo prático é prever acessos a um site de e-commerce com base em investimento em publicidade.

### O que é o coeficiente de determinação (R²) e qual é sua importância?
O coeficiente de determinação (R²) é uma medida que indica o quão bem o modelo de regressão se ajusta aos dados. Ele é importante porque ajuda a avaliar a precisão do modelo, indicando a proporção da variabilidade dos dados que é explicada pelo modelo.

### Como a regressão linear múltipla difere da regressão linear simples?
A regressão linear múltipla expande o modelo para incluir múltiplas variáveis independentes, ao contrário da regressão linear simples que considera apenas uma variável independente. Isso adiciona complexidade ao modelo, mas permite capturar relações mais complexas entre variáveis.

### Por que é importante validar modelos de regressão com dados históricos?
Validar modelos de regressão com dados históricos é crucial para garantir a precisão das previsões. Isso ajuda a identificar se o modelo é capaz de generalizar bem para novos dados e se as previsões são confiáveis em cenários reais.

### Quais ferramentas são recomendadas para implementar modelos de regressão?
O módulo enfatiza a importância de explorar diferentes bibliotecas para implementar modelos de regressão, como as disponíveis em Python. Familiarizar-se com várias ferramentas é uma habilidade frequentemente exigida no mercado de trabalho, permitindo flexibilidade e eficiência na modelagem.

### Como a matriz de correlação é utilizada na regressão linear múltipla?
A matriz de correlação é usada para identificar relações entre variáveis na regressão linear múltipla. Ela ajuda a entender quais variáveis independentes têm maior impacto na variável dependente, auxiliando na seleção de variáveis para o modelo.

# Projeto de Parceria

### Aula 1

Construa novas oportunidades
Neste curso, você tem mais uma oportunidade de aprendizagem: participar do projeto de parceria! A EBAC tem feito parcerias com empresas consolidadas no mercado a fim de garantir que seus estudantes vivenciem experiências práticas por meio da criação de um novo projeto para o seu futuro portfólio e assim desmistificar a complexidade das atividades cotidianas do profissional da área. Os trabalhos mais surpreendentes serão verificados pela empresa e receberão o feedback profissional. Os estudantes que se destacarem poderão participar do processo seletivo da empresa parceira.

Trata-se de uma atividade obrigatória caso você esteja no Programa de Empregabilidade e que fará total diferença para o seu desempenho profissional! Que tal conhecer com mais detalhes nosso projeto de parceria?

### Aula 2

Vamos avançar com o projeto de parceria? Lembre-se de que se trata de uma atividade obrigatória caso esteja no Programa de Empregabilidade. Até o momento, você já adquiriu conhecimento sobre variáveis e tipos de dados, estruturas de dados e metodologia para análise de dados. A partir disso, atente-se às orientações a seguir.  

Antes de começar

Relembre o briefing:  Encontrar uma problemática da vida real que possa ser solucionada através de análise de dados e machine learning. A ideia do projeto é explanar e justificar a relevância do uso de dados para encontrar a solução. 

Neste momento, você irá determinar a problemática e iniciar a coleta de informações. Esse projeto deve ser feito em formato Python através do Jupyter Notebook.

Passo a passo

Com base no briefing, defina qual a problemática do dia a dia escolhida para desenvolver este projeto. 
Levante as fontes de dados públicas e não confidenciais para a coleta de informações. 

### Aula 3

Vamos avançar com o projeto de parceria?
Antes de começar

Até aqui, você já definiu a problemática e levantou as fontes de informação não confidenciais sobre ela. Agora, é hora de analisar os dados e levantar possíveis soluções. 

Passo a passo

Efetue a análise exploratória dos dados levantando os principais pontos relevantes para a solução do problema.

Justifique os principais pontos escolhidos para a solução do problema e descreva sua relevância. 

### Aula 4

Até o momento, você já adquiriu conhecimento sobre árvores, GitHub, agregações e operações em grupo, gráficos interativos e combinação de modelos. A partir disso, atente-se às orientações a seguir.  

Antes de começar

Com os dados analisados e soluções definidas, agora você pode construir o modelo estatístico através de técnicas de modelagem para demonstrar os resultados. 

Passo a passo

Construa um aprendizado de máquina ou modelo estatístico.

Avalie o modelo estatístico através de técnicas de modelagem aprendidas ao longo do curso.

### Aula 5

Vamos finalizar o projeto de parceria?
Você já adquiriu conhecimento sobre combinação de modelos, cálculo, Algelin, PCA, K-means, probabilidade, inferência e regressão,. A partir disso, atente-se às orientações a seguir.  

Antes de começar

Com todos os dados levantados, agora é hora de estruturar a visualização dos resultados e fazer a entrega do projeto. 

Passo a passo

Apresente uma visualização de dados com os resultados obtidos.

Faça a entrega dessa visualização de dados via GitHub. Na sua entrega deve conter uma documentação em formato Markdown ou PDF com os tópicos: Coleta de dados, modelagem e conclusões.

# Glossário

### Ajuste de Hiperparâmetros
Processo de otimização dos parâmetros de um modelo de aprendizado de máquina que não são aprendidos a partir dos dados, mas configurados antes do treinamento para melhorar o desempenho do modelo.

### Análise Exploratória de Dados (EDA)
Técnica utilizada para resumir as principais características de um conjunto de dados, frequentemente com métodos visuais, antes de aplicar modelos de aprendizado de máquina.

### Desbalanceamento de Classes
Situação em que as classes de uma variável alvo em um conjunto de dados não estão igualmente representadas, o que pode afetar o desempenho de modelos de aprendizado de máquina.

### Divisão Simples
Método de avaliação de modelos onde o conjunto de dados é dividido em duas partes: uma para treinamento e outra para teste, sem a repetição de testes em diferentes subconjuntos.

### Generalização
Capacidade de um modelo de aprendizado de máquina de performar bem em dados novos e não vistos, além dos dados de treinamento.

### Overfitting
Fenômeno onde um modelo de aprendizado de máquina se ajusta muito bem aos dados de treinamento, mas falha em generalizar para dados novos.

### Parâmetros de Busca
Processo de encontrar os melhores valores para os hiperparâmetros de um modelo, geralmente através de técnicas como grid search ou random search.

### Previsão
Ação de usar um modelo de aprendizado de máquina treinado para estimar ou prever o valor de uma variável alvo em novos dados.

### Robustez
Capacidade de um modelo de aprendizado de máquina de manter um bom desempenho mesmo quando há variações nos dados de entrada ou condições de teste.

### Validação Cruzada (Cross Validation)
Técnica de avaliação de modelos que divide o conjunto de dados em múltiplos subconjuntos, permitindo testes repetidos em diferentes partes para obter uma avaliação mais confiável do desempenho do modelo.

# Glossário

### Componentes Principais
Variáveis não correlacionadas resultantes da transformação de um conjunto de dados original, utilizadas para simplificar a análise de dados em técnicas de redução de dimensionalidade como o PCA.

### Covariância
Uma medida estatística que indica o grau em que duas variáveis variam juntas. No contexto do PCA, a matriz de covariância é usada para identificar a direção dos componentes principais.

### Dimensionalidade
Refere-se ao número de variáveis ou características em um conjunto de dados. A redução de dimensionalidade é um processo para simplificar a análise de dados, reduzindo o número de variáveis.

### Maldição da Dimensionalidade
Um fenômeno em que o aumento do número de dimensões em um conjunto de dados dificulta a análise, tornando mais difícil identificar padrões e aumentando o risco de sobreajuste.

### Matriz de Correlação
Uma tabela que mostra o coeficiente de correlação entre todas as possíveis combinações de variáveis em um conjunto de dados, usada no PCA para entender as relações entre variáveis.

### Padronização
O processo de normalizar os dados para que tenham média zero e variância um, essencial no PCA para garantir que todas as variáveis contribuam igualmente para a análise.

### Projeção
A transformação dos dados originais em um novo espaço dimensional formado pelos componentes principais selecionados, facilitando a visualização e interpretação dos dados.

### Redução de Dimensionalidade
Uma técnica usada para diminuir o número de variáveis em um conjunto de dados, mantendo a maior parte da informação relevante, como feito no PCA.

### Variabilidade
A medida de quanto os dados variam em relação à média. No PCA, a variabilidade explicada pelos componentes principais é usada para determinar quantos componentes são necessários.

### Variáveis Correlacionadas
Variáveis que têm uma relação estatística entre si. O PCA transforma essas variáveis em componentes principais que são não correlacionados, simplificando a análise.

# Glossário

### Cross Validation
Uma técnica de validação de modelo que divide o conjunto de dados em múltiplas partes chamadas "folds", permitindo que o modelo seja treinado e testado em diferentes subconjuntos dos dados para garantir uma avaliação mais robusta e confiável.

### F1 Score
Uma métrica de desempenho que combina precisão e recall em um único valor, útil para avaliar modelos de classificação, especialmente em conjuntos de dados desbalanceados.

### Fold
Uma das partes em que o conjunto de dados é dividido durante a validação cruzada, onde cada fold é usado como conjunto de teste enquanto os restantes são usados para treino.

### Generalização
A capacidade de um modelo de aprendizado de máquina de performar bem em dados não vistos, ou seja, dados que não foram usados durante o treino do modelo.

### Hiperparâmetros
Parâmetros de um modelo de aprendizado de máquina que são definidos antes do processo de treinamento e influenciam o comportamento do modelo, como o número de folds na validação cruzada.

### K-Fold
Um tipo específico de validação cruzada onde o conjunto de dados é dividido em K partes, ou "folds", e o modelo é treinado e testado K vezes, cada vez usando um fold diferente como conjunto de teste.

### Overfitting
Um problema em modelos de aprendizado de máquina onde o modelo se ajusta muito bem aos dados de treino, mas falha em generalizar para novos dados, frequentemente mitigado por técnicas como a validação cruzada.

### Regressão Logística
Um modelo de classificação usado para prever a probabilidade de uma variável dependente binária, frequentemente utilizado em validação cruzada para avaliar sua performance em diferentes subconjuntos de dados.

### Subconjunto
Uma parte do conjunto de dados original usada em técnicas de validação cruzada para treino ou teste do modelo, permitindo uma avaliação mais completa do desempenho do modelo.

### Treino e Teste
A divisão tradicional de dados em dois conjuntos, onde o conjunto de treino é usado para ajustar o modelo e o conjunto de teste é usado para avaliar seu desempenho, uma abordagem que pode ser aprimorada pela validação cruzada.

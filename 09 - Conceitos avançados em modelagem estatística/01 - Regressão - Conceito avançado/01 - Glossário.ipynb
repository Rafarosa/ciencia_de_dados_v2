# Glossário

## Elastic Net
Uma técnica de regularização que combina as penalizações L1 (Lasso) e L2 (Ridge), oferecendo uma abordagem flexível para modelagem de dados complexos, especialmente útil quando há correlações entre variáveis.

## Multicolinearidade
Uma situação em que duas ou mais variáveis independentes em um modelo de regressão estão altamente correlacionadas, o que pode distorcer os resultados da análise e é frequentemente abordada com técnicas de regularização.

## PolynomialFeatures
Uma ferramenta da biblioteca SciKit-Learn em Python que transforma variáveis de entrada em suas potências, permitindo a criação de modelos de regressão polinomial que podem capturar relações não lineares nos dados.

## Regressão Lasso
Uma técnica de regularização que utiliza penalização L1 para simplificar modelos de regressão, eliminando variáveis menos relevantes e ajudando a prevenir o overfitting.

## Regressão Polinomial
Uma extensão da regressão linear que transforma variáveis de entrada em potências para ajustar curvas aos dados, permitindo capturar relações complexas que não seguem um padrão linear.

## Regressão Ridge
Uma técnica de regularização que utiliza penalização L2 para reduzir os coeficientes de um modelo de regressão, sem eliminá-los, sendo útil em situações de multicolinearidade.

## R² (R-quadrado)
Uma métrica de avaliação que indica a proporção da variância nos dados que é explicada pelo modelo de regressão, utilizada para avaliar a eficácia do ajuste do modelo.

## Regularização
Um conjunto de técnicas utilizadas para otimizar modelos de regressão, prevenindo o overfitting ao adicionar uma penalização aos coeficientes do modelo.

## Seleção de Variáveis
O processo de identificar e selecionar as variáveis mais relevantes para um modelo de regressão, frequentemente realizado por técnicas como a Regressão Lasso.

## Valor de Alfa
Um parâmetro que controla a força da penalização em técnicas de regularização, como Lasso e Ridge, influenciando o equilíbrio entre ajuste do modelo e simplicidade.

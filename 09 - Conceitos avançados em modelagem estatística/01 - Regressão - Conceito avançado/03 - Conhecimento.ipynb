# Conhecimento

## O que é regressão polinomial e quando ela deve ser usada?
A regressão polinomial é uma extensão da regressão linear que transforma variáveis de entrada em potências para ajustar curvas aos dados, em vez de linhas retas. Ela é útil quando a relação entre as variáveis não segue um padrão linear, permitindo capturar relações complexas nos dados.

## Como a regressão polinomial é implementada em Python?
Em Python, a regressão polinomial pode ser implementada usando a biblioteca SciKit-Learn. A transformação das variáveis de entrada em suas potências é feita com o `PolynomialFeatures`, e diferentes graus de polinômio podem ser testados para melhorar o ajuste do modelo.

## O que é a métrica R² e como ela é utilizada na regressão?
A métrica R², ou coeficiente de determinação, é utilizada para avaliar a eficácia de um modelo de regressão. Ela indica a proporção da variância dos dados que é explicada pelo modelo. Um R² mais próximo de 1 sugere um melhor ajuste aos dados.

## Quais são as técnicas de regularização abordadas no módulo?
O módulo abordou as técnicas de regularização Lasso, Ridge e Elastic Net. A Regressão Lasso realiza seleção de variáveis, eliminando as menos relevantes. A Regressão Ridge utiliza uma penalização L2 para reduzir os coeficientes sem eliminá-los, e o Elastic Net combina as penalizações L1 e L2, oferecendo uma abordagem flexível para modelagem de dados complexos.

## Como a regularização ajuda a prevenir o overfitting?
A regularização ajuda a prevenir o overfitting ao adicionar uma penalização aos coeficientes do modelo, o que evita que eles se ajustem excessivamente aos dados de treinamento. Isso resulta em um modelo mais simples e generalizável para novos dados.

## O que é multicolinearidade e como ela afeta os modelos de regressão?
A multicolinearidade ocorre quando duas ou mais variáveis independentes em um modelo de regressão estão altamente correlacionadas. Isso pode afetar a estabilidade dos coeficientes do modelo e dificultar a interpretação dos resultados. Técnicas de regularização, como Ridge e Elastic Net, podem ajudar a mitigar esse problema.

## Qual é a diferença entre as técnicas de regularização Lasso e Ridge?
A principal diferença entre Lasso e Ridge é a forma de penalização. Lasso utiliza uma penalização L1, que pode eliminar completamente variáveis menos relevantes, enquanto Ridge utiliza uma penalização L2, que reduz os coeficientes sem eliminá-los. Lasso é útil para seleção de variáveis, e Ridge é eficaz em situações de multicolinearidade.

## Como escolher o valor de alfa na regularização?
O valor de alfa controla a força da penalização na regularização. Ele pode ser escolhido através de validação cruzada, testando diferentes valores e selecionando aquele que resulta no melhor desempenho do modelo em dados de validação. Um alfa maior implica em uma penalização mais forte.

## Por que é importante testar diferentes modelos de regressão?
Testar diferentes modelos de regressão é importante para determinar qual modelo oferece o melhor ajuste aos dados. Diferentes modelos podem capturar diferentes padrões nos dados, e a comparação entre eles ajuda a escolher o mais adequado para o problema em questão.

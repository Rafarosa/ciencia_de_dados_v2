# Glossário

### Grid Search
Técnica de otimização de modelos que testa todas as combinações possíveis de um conjunto de hiperparâmetros para encontrar a configuração que proporciona o melhor desempenho.

### Hiperparâmetros
Parâmetros que são definidos antes do treinamento de um modelo de machine learning e que controlam o processo de aprendizagem, como a profundidade de uma árvore de decisão ou o valor de C na regressão logística.

### Overfitting
Fenômeno em que um modelo de machine learning se ajusta muito bem aos dados de treino, mas tem um desempenho ruim em dados de teste devido à sua complexidade excessiva.

### Padronização
Processo de escalonar os dados para que tenham média zero e desvio padrão um, frequentemente utilizado para melhorar o desempenho de modelos de machine learning.

### Penalty
Termo utilizado na regressão logística para regularização, ajudando a prevenir o overfitting ao adicionar uma penalidade à complexidade do modelo.

### SMOTE
Técnica de balanceamento de dados que cria exemplos sintéticos da classe minoritária para lidar com problemas de desbalanceamento em conjuntos de dados.

### Solver
Algoritmo utilizado para otimizar a função de custo na regressão logística, influenciando a velocidade e a precisão do treinamento do modelo.

### StandardScaler
Ferramenta de padronização de dados que transforma os dados para que tenham média zero e desvio padrão um, essencial para algoritmos que dependem da escala dos dados.

### Tree Depth
Profundidade de uma árvore de decisão, que determina o número máximo de divisões do nó raiz até a folha mais distante, influenciando a complexidade do modelo.

### Variáveis Categóricas
Variáveis que representam categorias ou grupos e que precisam ser codificadas numericamente para serem utilizadas em modelos de machine learning.

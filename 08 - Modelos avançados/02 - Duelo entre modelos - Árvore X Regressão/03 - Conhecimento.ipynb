# Conhecimento

### Quais são os pontos fortes e fracos da árvore de decisão e da regressão logística?
A árvore de decisão inicialmente apresenta um desempenho superior, mas pode sofrer de overfitting. A regressão logística, por outro lado, mostra melhorias significativas após o balanceamento e padronização dos dados, mantendo uma acurácia consistente entre os dados de treino e teste.

### Qual é a importância das otimizações realizadas para o desempenho dos modelos?
As otimizações são cruciais para melhorar o desempenho dos modelos, garantindo que eles sejam mais adequados para os dados específicos analisados e possam generalizar melhor para novos dados.

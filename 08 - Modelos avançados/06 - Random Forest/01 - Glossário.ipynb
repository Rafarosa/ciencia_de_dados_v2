# Glossário

### Criterion
Um hiperparâmetro no Random Forest que define a função a ser usada para medir a qualidade de uma divisão em uma árvore de decisão. Exemplos incluem o índice Gini e a entropia.

### Entropia
Uma medida de impureza ou desordem usada em árvores de decisão para determinar a qualidade de uma divisão dos dados. Quanto menor a entropia, mais homogêneo é o conjunto de dados.

### F1-score
Uma métrica de avaliação para modelos de classificação que considera tanto a precisão quanto o recall, sendo a média harmônica dessas duas métricas. É útil em cenários de classes desbalanceadas.

### Max_depth
Um hiperparâmetro que define a profundidade máxima de cada árvore no Random Forest. Controla o quão complexo cada árvore pode ser, ajudando a evitar o overfitting.

### Max_features
Um hiperparâmetro que determina o número máximo de características a serem consideradas para encontrar a melhor divisão em cada nó de uma árvore de decisão no Random Forest.

### Min_samples_leaf
Um hiperparâmetro que especifica o número mínimo de amostras que devem estar presentes em um nó folha de uma árvore de decisão no Random Forest.

### Min_samples_split
Um hiperparâmetro que define o número mínimo de amostras necessárias para dividir um nó em uma árvore de decisão no Random Forest.

### N_estimators
Um hiperparâmetro que indica o número de árvores de decisão a serem criadas no Random Forest. Um número maior de árvores pode aumentar a precisão, mas também o tempo de processamento.

### Overfitting
Um problema em modelos de machine learning onde o modelo se ajusta muito bem aos dados de treinamento, mas falha em generalizar para novos dados. O Random Forest ajuda a mitigar esse problema.

### ROC AUC
Uma métrica de avaliação para modelos de classificação que representa a área sob a curva ROC (Receiver Operating Characteristic), indicando a capacidade do modelo de distinguir entre classes.

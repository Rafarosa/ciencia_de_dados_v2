# Conhecimento

### O que é acurácia e por que pode ser enganosa em classes desbalanceadas?
A acurácia é a proporção de previsões corretas feitas por um modelo em relação ao total de previsões. Ela pode ser enganosa em classes desbalanceadas porque um modelo pode prever corretamente a classe majoritária muitas vezes, resultando em alta acurácia, mas falhar em prever a classe minoritária, que pode ser a mais importante.

### Qual é a diferença entre precisão e recall?
Precisão é a proporção de verdadeiros positivos entre todas as previsões positivas feitas pelo modelo. Recall é a proporção de verdadeiros positivos entre todos os casos reais positivos. Enquanto a precisão foca na qualidade das previsões positivas, o recall foca na capacidade do modelo de detectar todos os casos positivos.

### O que é o F1 Score e quando ele é útil?
O F1 Score é a média harmônica entre precisão e recall. Ele é útil em situações onde há um trade-off entre precisão e recall, fornecendo uma única métrica que considera ambos. É especialmente valioso quando as classes são desbalanceadas.

### Quais são as principais métricas para avaliar modelos de regressão?
As principais métricas para avaliar modelos de regressão incluem o coeficiente de determinação (R²), o Erro Médio Absoluto (MAE) e o Erro Quadrático Médio (MSE). O R² mede a proporção da variabilidade dos dados explicada pelo modelo, o MAE fornece uma medida clara do erro médio, e o MSE penaliza erros maiores, sendo mais sensível a outliers.

### O que são a curva ROC e a AUC?
A curva ROC (Receiver Operating Characteristic) ilustra a capacidade de um modelo de classificação binária em distinguir entre classes positivas e negativas. A AUC (Area Under the Curve) resume essa capacidade em uma única métrica, onde um valor mais próximo de 1 indica melhor desempenho do modelo.

### Qual é a diferença entre hiperparâmetros e parâmetros?
Hiperparâmetros são configurações externas ao modelo que controlam o comportamento do algoritmo, como a taxa de aprendizado ou o número de árvores em uma floresta aleatória. Parâmetros são valores internos ajustados pelo modelo durante o treinamento, como os coeficientes em uma regressão linear.

### Por que é importante ajustar hiperparâmetros?
Ajustar hiperparâmetros é importante para melhorar o desempenho do modelo e evitar overfitting. Hiperparâmetros bem ajustados podem levar a um modelo mais preciso e generalizável, enquanto ajustes inadequados podem resultar em um modelo que não captura bem os padrões dos dados.

### O que são GridSearchCV e RandomizedSearchCV?
GridSearchCV e RandomizedSearchCV são técnicas para otimizar a seleção de hiperparâmetros. GridSearchCV realiza uma busca exaustiva sobre um espaço especificado de hiperparâmetros, enquanto RandomizedSearchCV amostra aleatoriamente um número fixo de combinações de hiperparâmetros, sendo mais eficiente em termos de tempo.

### Como o R² pode ser limitado em casos de overfitting ou muitos outliers?
O R² pode ser limitado em casos de overfitting porque um modelo que se ajusta muito bem aos dados de treinamento pode ter um R² alto, mas não generalizar bem para novos dados. Em casos com muitos outliers, o R² pode ser enganoso porque esses pontos extremos podem distorcer a medida de variabilidade explicada pelo modelo.

### Por que é importante usar múltiplas métricas para avaliar modelos?
Usar múltiplas métricas para avaliar modelos é importante porque cada métrica fornece uma perspectiva diferente sobre o desempenho do modelo. Algumas métricas podem ser mais sensíveis a certos aspectos dos dados, como desbalanceamento de classes ou presença de outliers, e uma avaliação abrangente ajuda a garantir que o modelo seja robusto e eficaz em diferentes cenários.

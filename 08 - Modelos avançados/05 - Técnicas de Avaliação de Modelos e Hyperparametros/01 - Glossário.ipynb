# Glossário

### AUC (Área sob a Curva)
Uma métrica que resume a capacidade de um modelo de classificação binária em distinguir entre classes positivas e negativas. É calculada como a área sob a curva ROC.

### Curva ROC (Receiver Operating Characteristic)
Um gráfico que ilustra a capacidade de um modelo de classificação binária em distinguir entre classes positivas e negativas, plotando a taxa de verdadeiros positivos contra a taxa de falsos positivos.

### Erro Médio Absoluto (MAE)
Uma métrica de avaliação de modelos de regressão que calcula a média dos erros absolutos entre as previsões do modelo e os valores reais.

### Erro Quadrático Médio (MSE)
Uma métrica de avaliação de modelos de regressão que calcula a média dos quadrados dos erros entre as previsões do modelo e os valores reais, penalizando erros maiores.

### F1 Score
Uma métrica que combina precisão e recall, calculada como a média harmônica dessas duas métricas, útil em situações de trade-off entre precisão e recall.

### GridSearchCV
Uma técnica de otimização de hiperparâmetros que realiza uma busca exaustiva sobre um espaço especificado de parâmetros, utilizando validação cruzada para avaliar o desempenho do modelo.

### Hiperparâmetros
Configurações externas ao modelo que controlam o comportamento do algoritmo, sendo fundamentais para ajustar o desempenho do modelo e evitar overfitting.

### RandomizedSearchCV
Uma técnica de otimização de hiperparâmetros que realiza uma busca aleatória sobre um espaço especificado de parâmetros, utilizando validação cruzada para avaliar o desempenho do modelo.

### Recall
Uma métrica de avaliação de modelos de classificação que mede a capacidade do modelo em identificar corretamente todos os casos positivos.

### R² (Coeficiente de Determinação)
Uma métrica de avaliação de modelos de regressão que mede a proporção da variabilidade dos dados explicada pelo modelo, sendo limitada em casos de overfitting ou muitos outliers.

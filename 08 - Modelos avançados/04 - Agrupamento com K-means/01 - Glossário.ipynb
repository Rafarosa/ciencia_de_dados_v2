# Glossário

### Algoritmo K-means
Um método de clustering que particiona um conjunto de dados em K grupos baseados em características semelhantes, utilizando centroides para representar cada cluster.

### Centroides
Pontos que representam o centro de um cluster no algoritmo K-means, usados para calcular a proximidade dos dados e determinar a qual cluster eles pertencem.

### Clustering
Técnica de aprendizado de máquina não supervisionado que agrupa dados em clusters com base em semelhanças intrínsecas, sem rótulos predefinidos.

### Distância Euclidiana
Uma medida de distância entre dois pontos em um espaço euclidiano, frequentemente usada no K-means para calcular a proximidade entre dados e centroides.

### Método do Cotovelo
Técnica utilizada para determinar o número ideal de clusters em um algoritmo de clustering, analisando a variação da soma dos quadrados dentro dos clusters.

### Outliers
Dados que se desviam significativamente dos outros pontos de um conjunto, podendo afetar a formação de clusters no K-means.

### Padronização de Dados
Processo de normalizar dados para que tenham uma média de zero e um desvio padrão de um, garantindo que todas as características contribuam igualmente para a formação de clusters.

### Segmentação de Mercado
Aplicação do clustering para dividir um mercado em grupos distintos de consumidores com base em características comuns, como comportamento de compra.

### Sensibilidade à Inicialização
Refere-se à dependência do algoritmo K-means na escolha inicial dos centroides, que pode influenciar o resultado final dos clusters formados.

### Visualização de Dados
Uso de representações gráficas para interpretar e comunicar a formação de clusters, facilitando a análise e compreensão dos padrões nos dados.

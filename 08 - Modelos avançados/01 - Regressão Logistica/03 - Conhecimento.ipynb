# Conhecimento

### O que é regressão logística e para que é utilizada?
A regressão logística é um modelo de classificação em aprendizado supervisionado utilizado para prever categorias, como "sim" ou "não". É amplamente aplicada em áreas como detecção de spam, modelagem de risco de crédito e diagnósticos médicos.

### Como a função sigmoide é utilizada na regressão logística?
A função sigmoide é central na regressão logística, pois converte variáveis em probabilidades entre zero e um, facilitando a classificação binária. Ela transforma uma combinação linear de variáveis em uma probabilidade.

### Quais são os passos principais no aprendizado da regressão logística?
Os passos principais incluem estimar os coeficientes, realizar previsões de probabilidade e avaliar a precisão das previsões com métricas como acurácia, recall, precisão e F1-score.

### Por que o pré-processamento de dados é importante na regressão logística?
O pré-processamento de dados é crucial para garantir que as features estejam na mesma escala, evitando vieses no modelo. A padronização é realizada utilizando o StandardScaler, transformando os dados para que cada feature tenha uma média de zero e um desvio padrão de um.

### O que é o balanceamento de dados e como é realizado?
O balanceamento de dados é uma técnica utilizada para lidar com desbalanceamentos nas classes, garantindo que o modelo não seja tendencioso para uma classe específica. Técnicas como o SMOTE são utilizadas para balancear os dados de treino.

### Como a regressão logística é implementada em Python?
A regressão logística é implementada em Python utilizando a biblioteca Scikit-learn. A classe LogisticRegression é importada e o modelo é ajustado com dados padronizados e balanceados.

### Quais métricas são utilizadas para avaliar o modelo de regressão logística?
As métricas utilizadas incluem precisão, recall, F1 score, além da curva ROC e AUC, que são ferramentas para avaliar o desempenho de modelos de classificação binária.

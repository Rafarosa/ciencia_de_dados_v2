# Glossário

### Acurácia
Uma métrica de avaliação que mede a proporção de previsões corretas em relação ao total de previsões feitas por um modelo de classificação.

### AUC (Área sob a Curva)
Uma métrica que representa a área sob a curva ROC, utilizada para avaliar a capacidade de um modelo de classificação em distinguir entre classes positivas e negativas.

### Descida do Gradiente
Um método de otimização utilizado para ajustar os coeficientes de um modelo, minimizando o erro entre as previsões do modelo e os valores reais.

### F1-score
Uma métrica de avaliação que combina precisão e recall em um único valor, sendo especialmente útil em problemas de classificação com classes desbalanceadas.

### Função Sigmoide
Uma função matemática que transforma uma combinação linear de variáveis em uma probabilidade, essencial para a classificação binária na regressão logística.

### LogisticRegression
Uma classe da biblioteca Scikit-learn em Python utilizada para implementar a regressão logística, permitindo o ajuste de modelos de classificação binária.

### Newton-Raphson
Um método numérico utilizado para encontrar os coeficientes que minimizam o erro em modelos de regressão logística, através de aproximações iterativas.

### Precision (Precisão)
Uma métrica de avaliação que mede a proporção de verdadeiros positivos em relação ao total de previsões positivas feitas por um modelo de classificação.

### Recall
Uma métrica de avaliação que mede a proporção de verdadeiros positivos em relação ao total de elementos que realmente pertencem à classe positiva.

### SMOTE (Synthetic Minority Over-sampling Technique)
Uma técnica de balanceamento de dados que cria exemplos sintéticos da classe minoritária para lidar com desbalanceamentos em conjuntos de dados de treino.

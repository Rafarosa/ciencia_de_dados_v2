# Glossário

### Amazon SageMaker
Uma plataforma de serviço em nuvem que fornece ferramentas para construir, treinar e implantar modelos de machine learning de forma escalável e automatizada.

### Grid Search
Uma técnica de otimização de hiperparâmetros que realiza uma busca exaustiva em um espaço especificado de parâmetros para encontrar a combinação que resulta no melhor desempenho do modelo.

### Google Cloud
Uma plataforma de computação em nuvem que oferece uma variedade de serviços, incluindo ferramentas para machine learning e análise de dados, permitindo a automação e escalabilidade de projetos de ciência de dados.

### Hiperparâmetros
Parâmetros externos ao modelo que são ajustados antes do processo de treinamento, influenciando diretamente o desempenho e a capacidade de generalização do modelo.

### K-means
Um algoritmo de agrupamento que particiona um conjunto de dados em K grupos, minimizando a variância dentro de cada grupo e maximizando a distância entre os grupos.

### Overfitting
Um problema em modelos de machine learning onde o modelo se ajusta excessivamente aos dados de treinamento, perdendo a capacidade de generalizar para novos dados.

### SMOTE (Synthetic Minority Over-sampling Technique)
Uma técnica de pré-processamento usada para balancear conjuntos de dados desbalanceados, criando exemplos sintéticos da classe minoritária para melhorar o desempenho do modelo.

### StandardScaler
Uma técnica de normalização que padroniza as características de um conjunto de dados, removendo a média e escalando para a variância unitária, melhorando a performance de alguns algoritmos de machine learning.

### Transformação de variáveis categóricas
O processo de converter variáveis categóricas em um formato numérico que pode ser utilizado por algoritmos de machine learning, como one-hot encoding.

### Variáveis independentes e dependentes
Em um conjunto de dados, as variáveis independentes são aquelas que servem como entrada para o modelo, enquanto a variável dependente é o resultado ou saída que o modelo tenta prever.

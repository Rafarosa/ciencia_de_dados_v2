# Glossário

### Dashboard
Uma interface visual que exibe métricas e dados de performance do modelo em tempo real, permitindo o monitoramento contínuo e a rápida identificação de problemas.

### Implementação em Produção
O processo de integrar e disponibilizar um modelo de machine learning em um ambiente real, onde ele pode ser acessado e utilizado por outros sistemas e aplicações.

### Logs
Registros detalhados das atividades e eventos que ocorrem durante a execução do modelo, usados para monitorar e diagnosticar problemas de performance.

### Manutenção Contínua
A prática de atualizar e ajustar o modelo regularmente para garantir que ele continue a fornecer resultados precisos e relevantes à medida que os dados e o ambiente mudam.

### Métricas de Performance
Indicadores quantitativos usados para avaliar a eficácia do modelo, como acurácia, precisão, recall e tempo de resposta.

### Monitoramento Contínuo
O processo de observar e analisar constantemente o desempenho do modelo para garantir que ele opere conforme o esperado e para detectar rapidamente quaisquer desvios.

### Refazer Treinamento
A ação de re-treinar o modelo com novos dados para capturar mudanças e garantir que ele continue a ser eficaz em suas previsões.

### Teste Rigoroso
A prática de submeter o modelo a uma série de testes intensivos para assegurar que ele funcione corretamente em produção e que todas as integrações estejam operando como esperado.

### Treinamento do Modelo
A fase em que o modelo aprende padrões a partir dos dados fornecidos, ajustando seus parâmetros para otimizar a performance.

### Validação do Modelo
O processo de testar o modelo com um conjunto de dados separado para garantir que ele generalize bem e não esteja superajustado aos dados de treinamento.
